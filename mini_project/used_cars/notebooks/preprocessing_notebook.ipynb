```python
# XGBoost can work with NaNs so we will do the following 


choosen_df['engine_l'] = np.where(choosen_df['engine_l_imputed']==1, np.nan, choosen_df['engine_l'])

choosen_df
```

```python
# 1) near-zero km on an old car: treat as missing, not as real data
mask_low = (choosen_df.kmDriven < 1000) & (choosen_df.Age >= 5)
mask_high = np.where(choosen_df.kmDriven / choosen_df.Age.clip(lower=1) < 60000 , 0, 1)


choosen_df.loc[np.logical_or(mask_low, mask_high), 'kmDriven'] = np.nan

choosen_df
```

```python
bm_group = choosen_df.groupby('brand_model')
bm_group_median_price = bm_group['AskPrice'].transform('median')
bm_group_count = bm_group['AskPrice'].transform('size')

askprice_median_ratio = choosen_df.AskPrice / bm_group_median_price
unbiased_sample = (bm_group_count >= 20)

low_mask = (unbiased_sample) & (askprice_median_ratio < 0.15) & (choosen_df.Age <=10)
high_mask = (unbiased_sample) & (askprice_median_ratio > 5)

choosen_df.loc[high_mask, 'AskPrice'] = (bm_group_median_price * 5)[high_mask]
choosen_df = choosen_df[~low_mask].reset_index(drop=True)
```

```python
from sklearn.model_selection import train_test_split, cross_val_score, KFold

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.preprocessing import OneHotEncoder, TargetEncoder
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from xgboost import XGBRegressor 
```

```python
y = np.log1p(choosen_df['AskPrice'])
X = choosen_df.drop(columns=['AskPrice', 'Brand', 'engine_l_imputed'], errors='ignore')

num_cols = ['Age', 'kmDriven', 'engine_l']          
bin_cols = ['Transmission', 'Owner']
cat_cols = ['FuelType']
te_cols  = ['brand_model']


X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
pre = ColumnTransformer([
    ('te',  TargetEncoder(cv=5, random_state=0), te_cols),
    ('bin', OneHotEncoder(drop='if_binary', sparse_output=False), bin_cols),
    ('ohe', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols),
    ('num', 'passthrough', num_cols),
])

model = Pipeline([
    ('pre', pre),
    ('xgb', XGBRegressor(n_estimators=600, learning_rate=0.05, max_depth=6,
                         subsample=0.8, colsample_bytree=0.8, random_state=0, n_jobs=-1)),
])
```

```python
cv_rmse = -cross_val_score(model, X_train, y_train, scoring='neg_root_mean_squared_error',
                           cv=KFold(5, shuffle=True, random_state=0))
print(f"CV RMSE (log scale): {cv_rmse.mean():.4f} ± {cv_rmse.std():.4f}")

model.fit(X_train, y_train)
pred = model.predict(X_test)
```

```python
print(f"Test RMSE (log): {np.sqrt(mean_squared_error(y_test, pred)):.4f}")
print(f"Test R2        : {r2_score(y_test, pred):.4f}")
print(f"Test MAE (INR) : {mean_absolute_error(np.expm1(y_test), np.expm1(pred)):,.0f}")
print(f"Median % error : {np.median(np.abs(np.expm1(pred) / np.expm1(y_test) - 1)) * 100:.1f}%")
```

```python
import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, RegressorMixin
```

```python
class GroupMedianRegressor(BaseEstimator, RegressorMixin):
    """Predicts the median target of each group (e.g. brand_model).
    Unseen groups fall back to the global training median."""

    def __init__(self, group_col='brand_model'):
        self.group_col = group_col

    def fit(self, X, y):
        y = pd.Series(np.asarray(y), index=X.index)
        self.group_medians_ = y.groupby(X[self.group_col]).median()
        self.global_median_ = float(y.median())
        return self

    def predict(self, X):
        return (X[self.group_col].map(self.group_medians_)
                                  .fillna(self.global_median_)
                                  .to_numpy())


```

```python
base = GroupMedianRegressor().fit(X_train, y_train)
base_pred = base.predict(X_test)

rmse = lambda a, b: np.sqrt(mean_squared_error(a, b))

print(f"GroupMedian  RMSE (log): {rmse(y_test, base_pred):.4f} | R2: {r2_score(y_test, base_pred):.4f} | "
      f"median % err: {np.median(np.abs(np.expm1(base_pred) / np.expm1(y_test) - 1)) * 100:.1f}%")
print(f"XGBoost      RMSE (log): {rmse(y_test, pred):.4f} | R2: {r2_score(y_test, pred):.4f} | "
      f"median % err: {np.median(np.abs(np.expm1(pred) / np.expm1(y_test) - 1)) * 100:.1f}%")


cv = KFold(5, shuffle=True, random_state=0)
base_cv = -cross_val_score(GroupMedianRegressor(), X_train, y_train,
                           scoring='neg_root_mean_squared_error', cv=cv)
print(f"GroupMedian CV RMSE: {base_cv.mean():.4f} ± {base_cv.std():.4f}")
```

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint, uniform

param_dist = {
    'xgb__n_estimators':     randint(300, 1500),
    'xgb__learning_rate':    uniform(0.01, 0.09),    # 0.01 - 0.10
    'xgb__max_depth':        randint(3, 10),
    'xgb__min_child_weight': randint(1, 12),
    'xgb__subsample':        uniform(0.6, 0.4),      # 0.6 - 1.0
    'xgb__colsample_bytree': uniform(0.5, 0.5),      # 0.5 - 1.0
    'xgb__gamma':            uniform(0, 0.5),
    'xgb__reg_alpha':        uniform(0, 2),
    'xgb__reg_lambda':       uniform(0.5, 5),
}

search = RandomizedSearchCV(
    model,                                           # your exact pipeline
    param_distributions=param_dist,
    n_iter=40,
    cv=KFold(5, shuffle=True, random_state=0),       # same CV as your baseline
    scoring='neg_root_mean_squared_error',           # same metric as your baseline
    random_state=0,
    n_jobs=1,                                        # XGBoost already uses all cores
    verbose=1,
)
search.fit(X_train, y_train)

print(f"baseline CV RMSE : {cv_rmse.mean():.4f} ± {cv_rmse.std():.4f}")
print(f"tuned    CV RMSE : {-search.best_score_:.4f}")
print({k.replace('xgb__', ''): (round(v, 4) if isinstance(v, float) else v)
       for k, v in search.best_params_.items()})

# one look at the test set, same metrics as before
tuned = search.best_estimator_
pred_t = tuned.predict(X_test)
print(f"\nTest RMSE (log): {np.sqrt(mean_squared_error(y_test, pred_t)):.4f}")
print(f"Test R2        : {r2_score(y_test, pred_t):.4f}")
print(f"Test MAE (INR) : {mean_absolute_error(np.expm1(y_test), np.expm1(pred_t)):,.0f}")
print(f"Median % error : {np.median(np.abs(np.expm1(pred_t) / np.expm1(y_test) - 1)) * 100:.1f}%")
```

```python
import joblib
from pathlib import Path

path = Path("../final_tuned_model")
path.mkdir(parents=True, exist_ok=True)

# (optional) silence the TargetEncoder warning without changing behavior
tuned = search.best_estimator_
tuned.set_params(pre__te=TargetEncoder(cv=KFold(5, shuffle=True, random_state=0)))

# retrain on all data and save
feature_cols = te_cols + bin_cols + cat_cols + num_cols
final_model = tuned.fit(X[feature_cols], y)
joblib.dump({'model': final_model, 'features': feature_cols}, path/'used_car_price_model.joblib')

# helper for new cars
def predict_price(df_new):
    return np.expm1(final_model.predict(df_new[feature_cols]))

# quick sanity check on one example
example = pd.DataFrame([{
    'brand_model': 'Maruti Suzuki,Swift', 'Age': 6, 'kmDriven': 60000, 'engine_l': 1.2,
    'Transmission': 'Manual', 'Owner': 'first', 'FuelType': 'Petrol'}])
print(f"{predict_price(example)[0]:,.0f}")
```
